```python
import re
import os
import xml.etree.ElementTree as ET
from natsort import natsorted # pip install natsort

def replace(fname, in_dir, out_dir):
    tree = ET.parse(os.path.join(in_dir, fname))
    root = tree.getroot()
    xml = ET.tostring(root ,encoding="iso-8859-1").decode(encoding="iso-8859-1") # For Python 2 or 3

    regex = 'id="S[0-9]*">\s*<FORM kindOf="original">(\d+)'

    instance = re.search(regex, xml)
    while instance:
        verse_number = instance.group(1)
        replacement = 'id="verse'+verse_number+'">\n\t\t<FORM kindOf="original">' 
        xml = re.sub(regex, replacement, xml, 1)
        instance = re.search(regex, xml)


    with open(os.path.join(out_dir, fname), "w", encoding='utf-8') as f:
        f.write(xml)
```
