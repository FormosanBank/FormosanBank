```python
import re
import os
import xml.etree.ElementTree as ET
from natsort import natsorted # pip install natsort


def fix(fname):
    tree = ET.parse(fname)
    root = tree.getroot()
    chapter = ET.fromstring("<TEXT></TEXT>")
    verse = False
    for elt in root:
        id = elt.attrib.get('id')
        if 'verse' in id:
            if verse:
                chapter.append(verse)
            verse = elt
        else:
            elt_text = ""
            for s in elt:
                elt_text = s.text
            for s in verse:
                s.text += " " + elt_text
    if verse:
        chapter.append(verse)
    
    out = 'combined-'+fname
    chapter = ET.ElementTree(chapter)
    chapter.write(out, xml_declaration=True, encoding="utf-8")

    return
    
```

```python

files = os.listdir()
files = natsorted(files)    #CORRECT numerical sort!
for file in files:
    if not ('combine' in str(file)):
        fix(file)
    print(file + " done")
```

```python
def list_sub(ls, sub, rep):
    new = []
    for l in ls:
        new.append(l.replace(sub, rep))
    return new
```

```python
def make_chapters(dir):
    i=0
    chapters = {i:[]}
    files = os.listdir(dir)
    files = natsorted(files)
    for file in files:
        # print("Processing file: " + file)
        with open(os.path.join(dir, file)) as f:
            lines = f.read()
        verses = lines.split("\n\n")
        for verse in verses:
            first = verse.split(" ")[0]

            if first == '1' or first == '1.':
                i+=1
                chapters.update({i:[verse]})
            else:
                chapters.get(i).append(verse)

    for key in chapters:
        chapters.update({key: list_sub(chapters.get(key), '\n', ' ')})
    

    for key in chapters:
        txt = chapters.get(key)
        new_txt = []
        i = -1
        versenum = 0
        for verse in txt:
            first = verse.split(" ")[0]
            first = first.replace('.', '')
            if first.isdigit():
                if int(first) == versenum:
                    verse = verse.replace(first, '')
                    new_txt[i] += verse
                else:
                    new_txt.append(verse)
                    versenum = int(first)
                    i+=1
            else:
                if 'Het' not in first:
                    new_txt[i] += verse
        chapters.update({key:new_txt})
    return chapters
```

```python
def add_xml(chapters, xml_dir):
    for key in chapters:
        txt = chapters.get(key)
        fname = 'chapter' + str(key)+'.xml'
        if fname in os.listdir(xml_dir):
            print(fname)
            chapter_xml = ET.parse(os.path.join(xml_dir, fname)).getroot()
            for verse in txt:
                elt = ET.fromstring("<TRANS></TRANS>")
                num = verse.split(" ")[0].replace(".", '')
                verse = verse.replace(str(num), '')
                elt.text = verse
                elt.set('LANG', 'nld')
                
                for sub in chapter_xml:
                    if sub.attrib.get('id') == 'verse'+num:
                        sub.append(elt)  
            out_file = os.path.join(xml_dir, 'trans-'+fname)
            out_xml = ET.ElementTree(chapter_xml)
            out_xml.write(out_file, xml_declaration=True, encoding="utf-8") 
        else:
            print(fname + " not found")
```

```python

chapters = make_chapters('../ocr-dutch-work/extraction/ocr_cleaned/matthew')

add_xml(chapters, './Matthew')
```
