```python
# Step 2: Import libraries
import pytesseract
from PIL import Image
import os
```

```python
def extract(title, lazy=True):

    input_folder_path = os.path.join('./book_scans_img', title)  # Replace 'your_folder_name' with your folder's name
    output_folder_path = os.path.join('./ocr_cleaned', title)  # Replace with desired output folder name

    ################################
    # Step 5: Create the output folder if it doesn't exist
    if lazy:
        if os.path.exists(output_folder_path):
            print("Text extraction skipped. Already completed.")
            return
        
    if not os.path.exists(output_folder_path):
        os.makedirs(output_folder_path)
    
    ################################
    # Step 6: Loop through all images in the input folder
    for filename in os.listdir(input_folder_path):
        if filename.endswith(('.png', '.jpg', '.jpeg', '.tiff', '.bmp', '.gif')):  # Process only image files
            image_path = os.path.join(input_folder_path, filename)
            image = Image.open(image_path)

            image = image.convert('L')  # Grayscale conversion
            image = image.point(lambda x: 0 if x < 128 else 255, '1')  # Thresholding for binary format


            # Extract text using Tesseract
            custom_config = '-l nldalt+enm+nld'  # Adjust based on layout

            extracted_text = pytesseract.image_to_string(image, config=custom_config)
            extracted_text = extracted_text.replace('ſ', 's')
            extracted_text = extracted_text.replace('v', 'b')
            extracted_text = extracted_text.replace('2', 'r')
            extracted_text = extracted_text.replace('z', 'r')

            # Save the extracted text to a separate text file
            text_file_name = os.path.splitext(filename)[0] + '.txt'  # Use image name for text file name
            text_file_path = os.path.join(output_folder_path, text_file_name)

            with open(text_file_path, 'w', encoding='utf-8') as file:
                file.write(extracted_text)

            print(f"Text extracted from {filename} saved as {text_file_name}")

    print("Text extraction complete. All results are saved in the output folder.")
    return
```

```python
path = './book_scans_img'
titles = os.listdir(path)

for title in titles:
    extract(title) 
print('\n\ndone with all extractions')
```
